```python
import sys; sys.path.append("..")
import pandas as pd
from src.data import load_data

X_train, X_test, y_train, y_test, labels = load_data()
cot = X_train.columns.tolist()

tr = X_train.copy(); tr["y"] = y_train.values
te = X_test.copy(); te["y"] = y_test.values

tr_dup = tr.drop_duplicates(subset=cot + ["y"])[cot + ["y"]]
gop = te.merge(tr_dup, on=cot + ["y"], how="left", indicator=True)
co_ban_sao = (gop["_merge"] == "both").to_numpy()

ten = te["y"].map(dict(enumerate(labels))).to_numpy()
bang = (pd.DataFrame({"lop": ten, "co_ban_sao": co_ban_sao})
        .groupby("lop")["co_ban_sao"].agg(["size", "sum", "mean"]).round(3))
bang.columns = ["so_mau_test", "so_mau_co_ban_sao_trong_train", "ti_le"]
bang.to_csv("../results/test/train_test_overlap.csv")
print("Ti le chung:", round(co_ban_sao.mean(), 3))
bang
```
