```python
import sys; sys.path.append("..")
from src.data import load_data
from src.config import ROOT, RANDOM_STATE

_, X_test, _, y_test, labels = load_data()

mau = X_test.groupby(y_test.values).sample(5, random_state=RANDOM_STATE)
mau.insert(0, "nhan_that", [labels[i] for i in y_test.loc[mau.index]])
mau = mau.sample(frac=1, random_state=RANDOM_STATE)
mau.to_csv(ROOT / "app" / "sample_flows.csv", index=False)

mau.groupby("nhan_that").head(1).to_csv(ROOT / "app" / "demo_upload.csv", index=False)
print(mau.shape)
```
