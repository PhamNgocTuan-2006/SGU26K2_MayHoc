```python
import sys
sys.path.append("..")

from src.data import load_data
from src.config import ROOT, RANDOM_STATE

# Tải dữ liệu train và danh sách tên nhãn
X_train, _, y_train, _, labels = load_data()

# Lấy ngẫu nhiên 5 mẫu cho mỗi lớp để làm tập dữ liệu mẫu cho app
mau_theo_lop = X_train.groupby(y_train.values).sample(
    n=5,
    random_state=RANDOM_STATE,
)

# Lấy nhãn số tương ứng với các dòng đã chọn rồi đổi sang tên nhãn chữ
y_mau = y_train.loc[mau_theo_lop.index]
ten_nhan_that = []
for idx_nhan in y_mau:
    ten_nhan_that.append(labels[idx_nhan])

# Chèn cột nhãn thật vào đầu bảng (vị trí cột 0)
mau_theo_lop.insert(0, "nhan_that", ten_nhan_that)

# Xáo trộn lại thứ tự các dòng để không bị xếp theo từng lớp
mau = mau_theo_lop.sample(frac=1, random_state=RANDOM_STATE)

# Lưu file mẫu ra thư mục app và in kích thước bảng
duong_dan_mau = ROOT / "app" / "sample_flows.csv"
mau.to_csv(duong_dan_mau, index=False)
print(mau.shape)
```
